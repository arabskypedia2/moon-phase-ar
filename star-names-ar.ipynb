# أسماء النجوم العربية في الفهرس الفلكي الحديث

دفتر تفاعلي يستعرض أشهر النجوم التي تحمل أسماء عربية الأصل في الفهارس الفلكية العالمية، مع حساب مواقعها الظاهرية.

المصدر والشروح الكاملة: [تراث أسماء النجوم — موسوعة السماء العربية](https://arabskypedia.com/heritage/star-names/)

تصفح [أطلس السماء التفاعلي](https://arabskypedia.com/atlas/) لرؤية هذه النجوم في مواقعها الليلة.

```python
stars = [
    ("الغول", "Algol", "β Persei", "عين رأس الغول في كوكبة حامل رأس الغول"),
    ("يد الجوزاء", "Betelgeuse", "α Orionis", "نجم أحمر عملاق في كتف الجبار"),
    ("رجل الجبار", "Rigel", "β Orionis", "ألمع نجوم كوكبة الجبار"),
    ("الدبران", "Aldebaran", "α Tauri", "عين الثور الحمراء"),
    ("فم الحوت", "Fomalhaut", "α Piscis Austrini", "من فم الحوت الجنوبي"),
    ("النسر الواقع", "Vega", "α Lyrae", "من النسر الواقع في كوكبة القيثارة"),
    ("النسر الطائر", "Altair", "α Aquilae", "من النسر الطائر في كوكبة العقاب"),
    ("ذنب الدجاجة", "Deneb", "α Cygni", "ذيل الدجاجة (الطائر)"),
    ("العيوق", "Capella", "α Aurigae", "من العنز في كوكبة ممسك الأعنة"),
    ("سهيل", "Canopus", "α Carinae", "ثاني ألمع نجوم السماء، لا يُرى من شمال أوروبا"),
]
print(f"{'الاسم العربي':<14} {'الاسم الدولي':<12} {'الترميز':<14} الوصف")
print('-' * 70)
for ar, en, des, note in stars:
    print(f"{ar:<14} {en:<12} {des:<14} {note}")
```
