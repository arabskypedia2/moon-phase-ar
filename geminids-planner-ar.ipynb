# مخطط رصد الجوزائيات 2026

دفتر تفاعلي لتخطيط مراقبة زخة شهب الجوزائيات، أقوى زخات السنة. الدليل الكامل بالمواعيد وأفضل أماكن المراقبة: [دليل شهب الجوزائيات 2026](https://arabskypedia.com/meteor-showers/geminids/)

مرتبط بـ [دليل زخات الشهب الرئيسي](https://arabskypedia.com/meteor-showers/) في موسوعة السماء العربية.


## لماذا الجوزائيات؟

- معدل يتجاوز 120 شهابًا في الساعة في الذروة
- شهب بطيئة نسبيًا وساطعة، مثالية للمبتدئين
- تنشط من مساء مبكر، بعكس معظم الزخات

حدّد الكوكبة المشعّة (الجوزاء) في سمائك عبر [أطلس السماء التفاعلي](https://arabskypedia.com/atlas/).


```python
from datetime import date, timedelta

peak = date(2026, 12, 14)  # ليلة الذروة المتوقعة
window = [peak + timedelta(days=d) for d in range(-2, 3)]
print('أفضل ليالي المراقبة حول الذروة:')
for d in window:
    mark = ' <-- الذروة' if d == peak else ''
    print(' ', d.isoformat(), mark)
```
