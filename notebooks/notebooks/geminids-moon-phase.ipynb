# طور القمر ليلة ذروة زخة شهب التوأميات 2026

توضح هذه الدفترية حساب طور القمر ونسبة إضاءته ليلة ذروة [زخة شهب التوأميات](https://arabskypedia.com/meteor-showers/geminids/) (13-14 ديسمبر 2026)، لأن ضوء القمر هو العامل الحاسم في عدد الشهب المرئية بالعين المجردة.

الدليل الكامل للزخة: [موسوعة السماء العربية — شهب التوأميات](https://arabskypedia.com/meteor-showers/geminids/)

```javascript
// باستخدام مكتبة moon-phase-ar (جافاسكربت)
// npm install moon-phase-ar
import { moonPhase } from 'moon-phase-ar';

const peakNight = new Date('2026-12-14T02:00:00+02:00');
const phase = moonPhase(peakNight);
console.log(`نسبة الإضاءة: ${(phase.illumination * 100).toFixed(1)}%`);
console.log(`اسم الطور: ${phase.nameAr}`);
```

## تفسير النتيجة

كلما قلّت نسبة إضاءة القمر زاد عدد الشهب الخافتة التي يمكن رؤيتها. عند الذروة يتجاوز المعدل 120 شهابًا في الساعة من سماء مظلمة بلا قمر.

- كل زخات الشهب السنوية بمواعيدها في [دليل زخات الشهب](https://arabskypedia.com/meteor-showers/)
